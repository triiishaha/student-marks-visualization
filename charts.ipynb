# Student Marks Visualization

A tour of common matplotlib chart types (pie, bar, scatter, histogram, line, box, area) applied to a small toy dataset of 5 students' marks and attendance.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## The data

```python
data = {'Student' : ['A','B','C','D','E'], 'Marks' : [85,78,92,70,88], 'Attendance':[90,85,95,80,92]}

df = pd.DataFrame(data)
df
```

## Pie chart — marks distribution

```python
plt.pie(df['Marks'], labels=df['Student'], autopct='%1.1f%%')
plt.title('Pie Chart - Marks Distribution')
plt.show()
```

## Bar chart — marks by student

```python
plt.bar(df['Student'], df['Marks'])
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Bar Chart - Marks Distribution')
plt.show()
```

## Scatter chart — attendance vs. marks

```python
plt.scatter(df['Attendance'], df['Marks'])
plt.xlabel('Attendance')
plt.ylabel('Marks')
plt.title('Scatter Chart - Attendance vs Marks')
plt.show()
```

## Histogram of marks

```python
plt.hist(df['Marks'])
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Histogram pf marks')
plt.show()
```

## Line chart — marks by student

```python
plt.plot(df['Student'], df['Marks'])
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Line Chart')
plt.show()
```

## Box plot of marks

```python
plt.boxplot(df['Marks'])

plt.ylabel('marks')
plt.title('Box plot of marks')
plt.show()
```

yellow line approx avg of box plot as almost all marks = 85

## Area chart — marks by student

```python
plt.fill_between(df['Student'], df['Marks'])
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Area Chart')
plt.show()
```
